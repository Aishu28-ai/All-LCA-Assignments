```python
#Program to check whether a triangle is right angled

#Define a function to check right angled triangle
def right_angled(a, b, c):

    #Arrange sides in ascending order
    sides=sorted([a, b, c])

    #Largest side is the hypotenuse
    a=sides[0]
    b=sides[1]
    c=sides[2]
    
    #Pythogoras theorm
    if a**2+b**2==c**2:
        return True
    else:
        return False

#Accept the lengths of three sides
a=float(input("Enter the first side:"))
b=float(input("Enter the second side:"))
c=float(input("Enter the third side:"))

#Check whether the sides can form a triangle
if a+b>c and a+c>b and b+c>a:

    #Call the function
    if right_angled(a, b, c):
        print("The triangle is a right-angled triangle.")
    else:
        print("The triangle is not a right-angled triangle.")
        
else:
    print("Invalid triangle. The sides do not form a triangle.")
```
