```python
#A program to check if a string contains only a-z, A-Z and 0-9

s=input("Enter a string:")  #get input

#Check each character
valid=True
for ch in s:
    if not(('a'<=ch<='z') or
           ('A'<=ch<='Z') or
           ('0'<=ch<='9')):
       valid=False
       break

if valid:
    print("Yes")
else:
    print("No")
    
```

```python
#Another Method

s=input("Enter a string:")
valid=True
i=0

while i<len(s):
    if(('A'<=s[i]<='Z') or
       ('a'<=s[i]<='z')):
        i=i+1
    elif '0'<=s[i]<='9':
        i=i+1
    else:
        valid=False
        break
if valid:
    print("Yes")
else:
    print("No")
    
```
